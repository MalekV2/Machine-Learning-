```python
import math as m

def bn(n):
    b0 = [1]
    for i in range(1,n+1):
        res = 0
        for j in range(0,i):
            res += (m.factorial(i-1)/(m.factorial(i-j-1)*m.factorial(j)))*b0[j]
        b0.append(res)
    return res

print(bn(4))
```
